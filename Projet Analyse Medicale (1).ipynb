```python
import matplotlib.pyplot as plt
df["Age"].hist()
plt.title("Distribution des ages")
plt.xlabel("Age")
plt.ylabel("Nombre de patients")
# Savoir la tranche d'age dominant, l'effectif des patients par tranche d'age(hist() )
```

```python
df["Diagnostic"].value_counts().plot.bar()
plt.title("Nombre de patient par maladie")
# Je veux savoir la maladie Dominant et la pathologie rares
```

```python
df.boxplot(column="Temperature")
plt.title("Distribution des Temperature")
# Je veux Detecter de fortes fievres, les valeurs inhabituelles, les erreurs potentiels, voir immédiatement les patients hors normes 
```

```python
plt.scatter(df["Age"], df["Tension_Systolique"])
plt.xlabel("Age")
plt.ylabel("Tension")
plt.title("Age vs Tension")
# Je veux voir la variation de la tension avec l'age
```

```python
df["Duree_Hospitalisation"].hist()
plt.title("Duree d'hospitalisation")
# Je veux voir les séjours court, cas lourds, Dispersion
```

```python
df["Temperature"].plot()
plt.title("Evolution des temperatures")
```

```python
df.groupby("Diagnostic")["Age"].mean().plot.bar()
# je veux savoir l'age moyen selon chaque maladie
```
